```python
def load_kaggle_dataset(dataset_name, file_name):
    import os, json, zipfile, pandas as pd

    kaggle_dir = os.path.expanduser("~/.kaggle")
    print(kaggle_dir)
    os.makedirs(kaggle_dir, exist_ok=True)
    if not os.path.exists(f"{kaggle_dir}/kaggle.json"):
        raise FileNotFoundError("Place kaggle.json in your notebook folder before calling this.")

    os.chmod(f"{kaggle_dir}/kaggle.json", 0o600)

    !kaggle datasets download -d {dataset_name}
    zip_path = dataset_name.split("/")[-1] + ".zip"
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall("data")
    
    return pd.read_csv(f"data/{file_name}")
```
